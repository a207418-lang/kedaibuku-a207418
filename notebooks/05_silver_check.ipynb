Compare the raw bronze `customers.csv` with the cleaned silver `customers.parquet`.


## What changed, and why

The customer table shrank from 515 to 500 rows because 15 exact duplicate rows were removed, which is the uniqueness fix from Lab 4. The columns changed for PDPA and consistency reasons: `full_name`, `phone` and `mykad` were dropped under data minimisation, `email` was replaced by `email_hash` (for counting unique customers), `email_masked` (for display) and the `email_missing` flag, `mykad` became a coarse `age_group` band so nobody can be re-identified, the 22 inconsistent `state` spellings were mapped to one label per state, and `signup_date` is now a real datetime instead of text.


```python
import pandas as pd

bronze = pd.read_csv("../data/lake/bronze/customers.csv")
silver = pd.read_parquet("../data/lake/silver/customers.parquet")

print("BRONZE rows:", len(bronze))
print(bronze.head())

print()
print("SILVER rows:", len(silver))
print(silver.head())

print()
print("Bronze columns:", list(bronze.columns))
print("Silver columns:", list(silver.columns))
```
